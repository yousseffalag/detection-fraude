```python
model = joblib.load('../../ml_models_1/fraud_model_new_XGB.pkl')
```

```python
X_test = pd.read_csv('../../data/X_test_raw.csv')
y_test = pd.read_csv('../../data/y_test_raw.csv')
```

```python
y_scores = model.predict_proba(X_test)[:, 1]
best_threshold = 0.5
y_pred_final = (y_scores >= best_threshold).astype(int)
```

```python
print(f1_score(y_test, y_pred_final))
print(accuracy_score(y_test, y_pred_final))
print(recall_score(y_test, y_pred_final))
print(precision_score(y_test, y_pred_final))
```

```python
import pandas as pd

# Charger le dataset
df = pd.read_csv('../../data/ai_ml_dataset.csv')

# Vérifier les colonnes disponibles
print(df.columns)

# ⚠️ Suppose que la colonne "isFraud" existe pour indiquer fraude (1) ou non (0)
# et que ton modèle utilise ces colonnes :
required_columns = [
    'type', 'amount', 'nameOrig', 'oldbalanceOrg', 'newbalanceOrig',
    'nameDest', 'oldbalanceDest', 'newbalanceDest', 'hour_of_day', 'day_of_week' , 'isFraud'
]

df['hour_of_day'] = df['step'] % 24
df['day_of_week'] = (df['step'] // 24) % 7

# Garder seulement les colonnes nécessaires
df = df[required_columns]

# Séparer fraude et non-fraude
fraud_df = df[df['isFraud'] == 1]
nonfraud_df = df[df['isFraud'] == 0]

print(f"Transactions frauduleuses dispo: {len(fraud_df)}")
print(f"Transactions non-frauduleuses dispo: {len(nonfraud_df)}")

# Prendre un échantillon équilibré de 500 fraudes et 500 non-fraudes
fraud_sample = fraud_df.sample(n=500, random_state=42)
nonfraud_sample = nonfraud_df.sample(n=500, random_state=42)

# Concaténer les deux
sample_df = pd.concat([fraud_sample, nonfraud_sample]).sample(frac=1, random_state=42).reset_index(drop=True)

# Sauvegarder pour test
sample_df.to_csv("../../data/transactions_sample_1000.csv", index=False)

print("✅ Fichier transactions_sample_1000.csv créé avec 1000 transactions (500 fraude + 500 non-fraude)")
```
